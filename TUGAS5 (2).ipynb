```python
import numpy as np
import matplotlib.pyplot as plt

# Data yang BENAR (sesuai soal)
L = np.array([0.20, 0.30, 0.40, 0.50, 0.60])
T = np.array([0.9324, 1.1447, 1.2816, 1.4185, 1.5492])

# Linearization
x = L
y = T**2

# Regresi linear
A = np.vstack([x, np.ones(len(x))]).T
m, c = np.linalg.lstsq(A, y, rcond=None)[0]

# Hitung g
g = (4 * np.pi**2) / m

print("Gradien (m) =", m)
print("Nilai g =", g, "m/s^2")

# Plot
plt.scatter(x, y, label="Data")
plt.plot(x, m*x + c, label="Regresi", color="red")
plt.xlabel("L (m)")
plt.ylabel("T^2 (s^2)")
plt.legend()
plt.grid()
plt.show()
```

```python
# Interpolasi polinom derajat 4
coef = np.polyfit(L, T**2, 4)
poly = np.poly1d(coef)

# Turunan polinom
dpoly = np.polyder(poly)

# Ambil titik tengah
L_mean = np.mean(L)

# Gradien di titik tersebut
slope = dpoly(L_mean)

# Hitung g
g_interp = (4 * np.pi**2) / slope

print("Gradien interpolasi =", slope)
print("Nilai g (interpolasi) =", g_interp, "m/s^2")

# Plot
L_plot = np.linspace(min(L), max(L), 100)
plt.scatter(L, T**2, label="Data")
plt.plot(L_plot, poly(L_plot), label="Interpolasi", color="green")
plt.xlabel("L (m)")
plt.ylabel("T^2 (s^2)")
plt.legend()
plt.grid()
plt.show()
```

```python
# Data contoh (bisa kamu sesuaikan kalau ada dari soal)
x = np.array([0, 1, 2, 3, 4, 5])
F = np.array([0, 2, 4, 6, 8, 10])

# Regresi linear (F = kx)
k, c = np.polyfit(x, F, 1)

print("Konstanta pegas k =", k)

# Interpolasi polinom derajat tinggi
coef_interp = np.polyfit(x, F, len(x)-1)
poly_interp = np.poly1d(coef_interp)

# Plot
x_plot = np.linspace(min(x), max(x), 100)

plt.scatter(x, F, label="Data")
plt.plot(x_plot, k*x_plot + c, label="Regresi", color="red")
plt.plot(x_plot, poly_interp(x_plot), label="Interpolasi", linestyle="--")
plt.xlabel("x")
plt.ylabel("F")
plt.legend()
plt.grid()
plt.show()
```
