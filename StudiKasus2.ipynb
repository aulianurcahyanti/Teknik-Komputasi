1. Pendahuluan

Atmosfer merupakan lapisan gas yang menyelimuti suatu planet dan memiliki peranan penting dalam berbagai fenomena fisika, seperti cuaca, perpindahan panas, serta distribusi tekanan udara. Salah satu karakteristik utama atmosfer adalah tekanan yang cenderung menurun seiring bertambahnya ketinggian. Fenomena ini terjadi karena semakin tinggi suatu lokasi, jumlah molekul udara yang berada di atas lokasi tersebut semakin sedikit sehingga gaya tekan yang diberikan oleh kolom udara menjadi lebih kecil. Pada planet Mars, kondisi atmosfer jauh lebih tipis dibandingkan atmosfer Bumi sehingga perubahan tekanan terhadap ketinggian menjadi salah satu parameter penting dalam studi atmosfer dan eksplorasi planet.

Informasi mengenai distribusi tekanan atmosfer Mars sangat diperlukan dalam berbagai aplikasi, seperti perancangan wahana antariksa, sistem pendaratan rover, serta analisis kondisi lingkungan di permukaan Mars. Oleh karena itu, diperlukan suatu model matematis yang dapat menggambarkan hubungan antara tekanan atmosfer dan ketinggian. Hubungan tersebut dapat diperoleh melalui persamaan hidrostatik yang dikombinasikan dengan hukum gas ideal. Namun, ketika suhu atmosfer berubah terhadap ketinggian, persamaan yang dihasilkan menjadi persamaan diferensial yang cukup sulit diselesaikan secara analitik.

Metode numerik merupakan salah satu pendekatan yang dapat digunakan untuk menyelesaikan permasalahan tersebut. Pada studi kasus ini digunakan metode Runge-Kutta Orde 4 (RK4) untuk menghitung perubahan tekanan atmosfer Mars dari permukaan hingga ketinggian 20 km. Metode RK4 dipilih karena memiliki tingkat akurasi yang lebih tinggi dibandingkan metode Euler serta banyak digunakan dalam penyelesaian persamaan diferensial pada berbagai permasalahan fisika dan teknik.

2. Metodologi

Tekanan atmosfer Mars terhadap ketinggian dapat dimodelkan menggunakan persamaan hidrostatik. Persamaan ini menyatakan bahwa perubahan tekanan terhadap ketinggian dipengaruhi oleh densitas udara dan percepatan gravitasi yang bekerja pada atmosfer. Secara matematis, persamaan hidrostatik dituliskan sebagai:

dP/dz = -ρg

dengan:

P = tekanan atmosfer (Pa)
ρ = densitas udara (kg/m³)
g = percepatan gravitasi Mars (3,71 m/s²)

Untuk menghubungkan densitas dengan tekanan digunakan Hukum Gas Ideal:

ρ = P/(RT)

dengan:

R = konstanta gas spesifik atmosfer Mars (188,92 J/kg·K)
T = suhu atmosfer (K)

Dengan mensubstitusikan persamaan gas ideal ke dalam persamaan hidrostatik diperoleh:

dP/dz = -(Pg)/(RT(z))

Pada penelitian ini digunakan model suhu atmosfer Mars yang berubah terhadap ketinggian sebagai berikut:

T(z) = 230 - 1,8z + 10e^(-1,5z)

Karena suhu berubah terhadap ketinggian, persamaan diferensial yang dihasilkan tidak mudah diselesaikan secara analitik. Oleh karena itu digunakan metode numerik Runge-Kutta Orde 4 (RK4) untuk memperoleh solusi pendekatan.

Metode RK4 bekerja dengan menghitung empat kemiringan (k1, k2, k3, dan k4) pada setiap langkah perhitungan. Hasil akhir diperoleh dari rata-rata tertimbang keempat nilai tersebut sehingga menghasilkan akurasi yang lebih baik dibandingkan metode Euler.

Parameter yang digunakan pada simulasi adalah:

Tekanan awal (P₀) = 610 Pa
Gravitasi Mars (g) = 3,71 m/s²
Konstanta gas Mars (R) = 188,92 J/kg·K
Interval ketinggian (h) = 1 km
Rentang ketinggian = 0 km hingga 20 km

Seluruh perhitungan dilakukan menggunakan codingan Python.

3. Penyelesaian Menggunakan Metode RK4

Persamaan diferensial yang akan diselesaikan adalah:

dP/dz = -(Pg)/(R(230 - 1,8z + 10e^(-1,5z)))

Misalkan:

f(z,P) = -(Pg)/(R(230 - 1,8z + 10e^(-1,5z)))

Diketahui kondisi awal:

P(0) = 610 Pa

dan ukuran langkah:

h = 1 km

Rumus Runge-Kutta Orde 4 yang digunakan adalah:

k1 = h × f(zn,Pn)

k2 = h × f(zn + h/2, Pn + k1/2)

k3 = h × f(zn + h/2, Pn + k2/2)

k4 = h × f(zn + h, Pn + k3)

Kemudian nilai tekanan pada langkah berikutnya dihitung menggunakan:

Pn+1 = Pn + (k1 + 2k2 + 2k3 + k4)/6

Iterasi Pertama

Diketahui:

z₀ = 0 km

P₀ = 610 Pa

Suhu pada z = 0 km:

T(0) = 230 - 1,8(0) + 10e⁰

T(0) = 240 K

Menghitung k1:

k1 = -(610 × 3,71)/(188,92 × 240)

k1 = -0,0499

Menghitung k2:

T(0,5) = 230 - 1,8(0,5) + 10e^(-0,75)

T(0,5) ≈ 233,82 K

k2 ≈ -0,0512

Menghitung k3:

k3 ≈ -0,0512

Menghitung k4:

T(1) = 230 - 1,8(1) + 10e^(-1,5)

T(1) ≈ 230,43 K

k4 ≈ -0,0520

Selanjutnya tekanan pada ketinggian 1 km dihitung sebagai:

P₁ = 610 + (-0,0499 + 2(-0,0512) + 2(-0,0512) + (-0,0520))/6

P₁ ≈ 609,95 Pa

Dengan demikian tekanan atmosfer Mars pada ketinggian 1 km adalah sekitar 609,95 Pa. Proses yang sama dilakukan secara berulang hingga mencapai ketinggian 20 km menggunakan program Python.

```python
import numpy as np
import matplotlib.pyplot as plt

g = 3.71
R = 188.92
h = 1
zmax = 20

def T(z):
    return 230 - 1.8*z + 10*np.exp(-1.5*z)

def f(z, P):
    return -(P*g)/(R*T(z))

z = np.arange(0, zmax+h, h)
P = np.zeros(len(z))

P[0] = 610

for i in range(len(z)-1):

    k1 = h*f(z[i], P[i])

    k2 = h*f(z[i] + h/2, P[i] + k1/2)

    k3 = h*f(z[i] + h/2, P[i] + k2/2)

    k4 = h*f(z[i] + h, P[i] + k3)

    P[i+1] = P[i] + (k1 + 2*k2 + 2*k3 + k4)/6

print("Ketinggian (km)   Tekanan (Pa)")
for i in range(len(z)):
    print(f"{z[i]:>5}            {P[i]:.4f}")

plt.figure(figsize=(8,5))
plt.plot(z, P, marker='o')
plt.xlabel('Ketinggian (km)')
plt.ylabel('Tekanan (Pa)')
plt.title('Perubahan Tekanan Atmosfer Mars terhadap Ketinggian')
plt.grid(True)
plt.show()
```

Tekanan tertinggi berada pada permukaan Mars dengan nilai awal sebesar 610 Pa, kemudian terus berkurang ketika ketinggian meningkat hingga 20 km.

Metode RK4 mampu memberikan hasil yang stabil dan konsisten pada setiap langkah perhitungan. Dibandingkan metode Euler, RK4 memiliki tingkat akurasi yang lebih tinggi karena mempertimbangkan empat estimasi kemiringan dalam satu langkah perhitungan.
