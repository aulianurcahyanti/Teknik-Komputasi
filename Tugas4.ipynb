Misalkan terdapat sistem rangkaian listrik dengan 4 arus yang tidak diketahui, yaitu:

I1​,I2​,I3​,I4​

Berdasarkan hukum Kirchhoff, diperoleh sistem persamaan linear:

12I1​−3I2​+1I3​+2I4​=10

−2I1​+9I2​−2I3​+1I4​=5

1I1​−2I2​+7I3​−1I4​=7

2I1​+1I2​−3I3​+10I4​=8


```python
def lu_decomposition(A):
    n = len(A)
    L = np.eye(n)
    U = A.copy().astype(float)

    for i in range(n):
        for j in range(i + 1, n):
            factor = U[j, i] / U[i, i]
            L[j, i] = factor
            U[j, i:] -= factor * U[i, i:]

    return L, U

def solve_lu(L, U, v):
    n = len(L)

    y = np.zeros(n)
    for i in range(n):
        y[i] = v[i] - np.dot(L[i, :i], y[:i])

    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        x[i] = (y[i] - np.dot(U[i, i+1:], x[i+1:])) / U[i, i]

    return x
```

```python
A = np.array([
    [12, -3, 1, 2],
    [-2, 9, -2, 1],
    [1, -2, 7, -1],
    [2, 1, -3, 10]
])

v = np.array([10, 5, 7, 8])

L, U = lu_decomposition(A)
sol_lu = solve_lu(L, U, v)

print("=== METODE LU ===")
print("Solusi:", sol_lu)
```

```python
def jacobi(A, v, x0, tol=1e-6, max_iter=100):
    n = len(v)
    x = x0.copy().astype(float)
    x_new = np.zeros(n)

    for k in range(max_iter):
        for i in range(n):
            s = sum(A[i, j] * x[j] for j in range(n) if i != j)
            x_new[i] = (v[i] - s) / A[i, i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k + 1

        x = x_new.copy()

    return x, max_iter
```

```python
x0 = np.zeros(4)

sol_jacobi, iter_j = jacobi(A, v, x0)

print("=== METODE JACOBI ===")
print("Solusi:", sol_jacobi)
print("Iterasi:", iter_j)
```

```python
def gauss_seidel(A, v, x0, tol=1e-6, max_iter=100):
    n = len(v)
    x = x0.copy().astype(float)

    for k in range(max_iter):
        x_old = x.copy()

        for i in range(n):
            s = sum(A[i, j] * x[j] for j in range(n) if i != j)
            x[i] = (v[i] - s) / A[i, i]

        if np.linalg.norm(x - x_old, ord=np.inf) < tol:
            return x, k + 1

    return x, max_iter
```

```python
sol_gs, iter_gs = gauss_seidel(A, v, x0)

print("=== METODE GAUSS-SEIDEL ===")
print("Solusi:", sol_gs)
print("Iterasi:", iter_gs)
```
