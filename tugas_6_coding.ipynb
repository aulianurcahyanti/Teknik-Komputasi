```python
# Tugas 6 No 1
# Metode Persegi Titik Tengah

def f(x):
    return 5 + 2*x**2

a = 0
b = 4
n = 4

dx = (b-a)/n

jumlah = 0

for i in range(n):
    xt = a + (i+0.5)*dx
    jumlah = jumlah + f(xt)

W = jumlah*dx

print("Hasil untuk n = 4")
print("W =", W)

# mencari nilai yang sudah konvergen

tol = 1e-14
selisih = 1

while selisih > tol:

    hasil_lama = W

    n = n*2
    dx = (b-a)/n

    jumlah = 0

    for i in range(n):
        xt = a + (i+0.5)*dx
        jumlah = jumlah + f(xt)

    W = jumlah*dx

    selisih = abs(W - hasil_lama)

print("\nHasil konvergen")
print("n =", n)
print("W =", W)
print("selisih =", selisih)
```

```python
# Tugas 6 No 2
# Metode Trapesium

def C(T):
    return 20 + 0.05*T**2

a = 0
b = 100
n = 4

dT = (b-a)/n

jumlah = C(a) + C(b)

for i in range(1,n):
    jumlah = jumlah + 2*C(a+i*dT)

Q = jumlah*dT/2

print("Hasil untuk n = 4")
print("Q =", Q)

tol = 1e-14
selisih = 1

while selisih > tol:

    hasil_lama = Q

    n = n*2
    dT = (b-a)/n

    jumlah = C(a) + C(b)

    for i in range(1,n):
        jumlah = jumlah + 2*C(a+i*dT)

    Q = jumlah*dT/2

    selisih = abs(Q - hasil_lama)

print("\nHasil konvergen")
print("n =", n)
print("Q =", Q)
print("selisih =", selisih)
```

```python
# Tugas 6 No 3
# Metode Simpson 1/3

def I(t):
    return 2 + 3*t + t**2

a = 0
b = 6
n = 6

dt = (b-a)/n

jumlah = I(a) + I(b)

for i in range(1,n):

    if i % 2 == 1:
        jumlah = jumlah + 4*I(a+i*dt)
    else:
        jumlah = jumlah + 2*I(a+i*dt)

q = jumlah*dt/3

print("Hasil untuk n = 6")
print("q =", q)

tol = 1e-14
selisih = 1

while selisih > tol:

    hasil_lama = q

    n = n*2
    dt = (b-a)/n

    jumlah = I(a) + I(b)

    for i in range(1,n):

        if i % 2 == 1:
            jumlah = jumlah + 4*I(a+i*dt)
        else:
            jumlah = jumlah + 2*I(a+i*dt)

    q = jumlah*dt/3

    selisih = abs(q - hasil_lama)

print("\nHasil konvergen")
print("n =", n)
print("q =", q)
print("selisih =", selisih)
```
